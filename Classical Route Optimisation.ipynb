```python
import math
import random
from itertools import permutations

cities = {
    "Lucknow": (0, 10),
    "Kanpur": (18, 35),
    "Ayodhya": (40, 38),
    "Gonda": (66, 19)
}

def distance(city1, city2):
    x1, y1 = cities[city1]
    x2, y2 = cities[city2]
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

def route_cost(route):
    total = 0
    for i in range(len(route) - 1):
        total += distance(route[i], route[i + 1])
    total += distance(route[-1], route[0])
    return total

def random_route(cities_list):
    route = cities_list.copy()
    random.shuffle(route)
    return route


def random_search(cities_list, trails=5000):
    best_route = None
    best_cost = float("inf")
    for _ in range(trails):
        route = random_route(cities_list)
        cost = route_cost(route)
        if cost < best_cost:
            best_cost = cost
            best_route = route
    return best_route, best_cost

cities_list = list(cities.keys())

best_route, best_cost = random_search(cities_list, trails=10000)

print("Best city route found:", best_route)
print("Cost:", best_cost)
print()

best_cost = float('inf')          # Reset!
best_route = None

for perm in permutations(cities_list):
    route = list(perm)
    cost = route_cost(route)
    if cost < best_cost:
        best_cost = cost
        best_route = route


print("True Optimal Route:", best_route)
print("True Optimal Cost:", best_cost)
        


```
